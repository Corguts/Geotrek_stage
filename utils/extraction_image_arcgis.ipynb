```python
import os, shutil, zipfile

from arcgis.gis import GIS
gis = GIS("home")

# OBLIGATOIRE : Indiquer ci-dessous l'ID de votre couche d'entités
FeatureLayerID = "xxxxxxxxxxxxxxxxxxxxxxxxxxxxxxxxx"
# OBLIGATOIRE : Indiquer l'index de la sous-couche dans la couche d'entités (par défaut 0)
SubLayerIndex = 0
# OPTION: Vous pouvez éventuellement modifier le répertoire cible du fichier photos.zip
TargetFolder = '/arcgis/home'

# NE PAS MODIFIER LE CODE CI-DESSOUS
def zipdir(path, ziph):
    # ziph is zipfile handle
    for root, dirs, files in os.walk(path):
        for file in files:
            ziph.write(os.path.join(root, file), 
                       os.path.relpath(os.path.join(root, file), 
                                       os.path.join(path, '..')))
ZipFileName = 'pieces_jointes.zip'          
if os.path.isfile(TargetFolder + '/' + ZipFileName):
    os.remove(TargetFolder + '/' + ZipFileName)

FeatureLayer = gis.content.get(FeatureLayerID)
layer = FeatureLayer.layers[SubLayerIndex]
TempPhotoFollder = TargetFolder + "/pieces_jointes"
layer.attachments.download(oid = None, save_path=TempPhotoFollder)

with zipfile.ZipFile(TargetFolder + '/' + ZipFileName, 'w') as zipf:
    zipdir(TempPhotoFollder, zipf)
    
shutil.rmtree(TempPhotoFollder)
```
